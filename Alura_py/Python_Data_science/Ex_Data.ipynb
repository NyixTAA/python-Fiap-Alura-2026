```python
def calcula_pontos(gols_marcados, gols_sofridos):
    pontuacao = 0
    for i in range(len(gols_marcados)):
        if gols_marcados[i] > gols_sofridos[i]:
            pontuacao += 3
        elif gols_marcados [i] == gols_sofridos [i]:
            pontuacao += 1
    total_jogos = len(gols_marcados)
    pontos_possiveis = total_jogos * 3
    aproveitamento = (pontuacao / pontos_possiveis) * 100
    return aproveitamento, pontuacao

aproveitamento, pontuacao = calcula_pontos([2, 1, 3, 1, 0], [1, 2, 2, 1, 3])
print(f'Pontuação do time: {pontuacao}\nAproveitamento do time: {aproveitamento:.2f}%')
```

```python
cidade = input('Digite para qual cidade gostaria de viajar (Salvador, Fortaleza, Natal ou Aracaju): ').upper()
while cidade not in ('SALVADOR','FORTALEZA','NATAL','ARACAJU'):
    cidade = input('Digite um destino válido (Salvador, Fortaleza, Natal ou Aracaju): ').upper()
dias = int(input('Digite quantas diárias gostaria: '))
distancias = [850, 800, 300, 550]
passeios = [200, 400, 250, 300]
gasolina = 5
km_l = 14

def gasto_hotel(dias):
    return 150 * dias

def gasto_gasolina(cidade):
    if cidade == 'SALVADOR':
        return (2 * distancias[0] / km_l) * gasolina
    elif cidade == 'FORTALEZA':
        return (2 * distancias[1] / km_l) * gasolina
    elif cidade == 'NATAL':
        return (2 * distancias[2] / km_l) * gasolina
    elif cidade == 'ARACAJU':
        return (2 * distancias[3] / km_l) * gasolina

def gasto_passeio(cidade, dias):
    if cidade == 'SALVADOR':
        return passeios[0] * dias
    elif cidade == 'FORTALEZA':
        return passeios[1] * dias
    elif cidade == 'NATAL':
        return passeios[2] * dias
    elif cidade == 'ARACAJU':
        return passeios[3] * dias

gastos = gasto_hotel(dias) + gasto_gasolina(cidade) + gasto_passeio(cidade, dias)



print(f'\nCom base nos gastos definidos, uma viagem de {dias} dias para {cidade.capitalize()} saindo de Recife custaria R$ {gastos:.2f}')
```

```python
def converte_lista(lista):
    try:
        nova_lista = [float(elemento) for elemento in lista]
    except Exception as e:
        print(type(e), f'Erro: {e}')
    else:
        return nova_lista
    finally:
        print('Fim da execução da função')
converte_lista([2,4,8,2])
```

```python
gabarito = ['D', 'A', 'B', 'C', 'A']
pontuacoes = []

def corretor(respostas: list):
    try:
            for teste in respostas:
                nota = 0
                for i in range(len(teste)):
                    if teste[i] not in ['A','B','C','D']:
                        raise ValueError(f'A alternativa {teste[i]} não é uma opção de alternativa válida')
                    elif teste[i] == gabarito[i]:
                        nota += 1
                pontuacoes.append(nota)
                    
    except Exception as e:
        print(type(e), f'Erro: {e}')
    
    else:
        return pontuacoes

corretor([['D', 'A', 'B', 'C', 'A'], ['C', 'A', 'A', 'C', 'A'], ['D', 'B', 'A', 'C', 'A']])
```

```python
def verificar(lista: list):
    try: 
        for palavra in lista:
            if (',' or '.' or '!' or '?') in palavra:
                raise ValueError(f'O texto apresenta pontuações na palavra "{palavra}".')
    except Exception as e:
        print(type(e), f'Erro: {e}')

    else:
        print('A lista está tratada.')

    finally: 
        print('Verificação concluída')

verificar(['Python', 'é', 'uma', 'linguagem,', 'de', 'programação', 'poderosa', 'versátil',

                  'e', 'fácil', 'de', 'aprender', 'utilizada', 'em', 'diversos', 'campos', 'desde',

                  'análise', 'de', 'dados', 'até', 'inteligência', 'artificial'])
```

```python
def divide_colunas(lista1: list, lista2: list):
    resultado = []
    if len(lista1) != len(lista2):
            raise ValueError('As duas listas devem ter o mesmo tamanho.')
        
    try:
        for num, den in zip(lista1, lista2):
                resultado.append(f'Razão entre {num} e {den} = {(num / den):.2f}')
    except ZeroDivisionError as e:
        print(type(e), f'O denominador não pode ser zero.')

    else:
        return resultado
    finally:
        print('Finalizado!')


resultado = divide_colunas([100, 120, 140, 160, 180],[20, 25, 30, 35, 2])
print(*resultado, sep='\n')
```
