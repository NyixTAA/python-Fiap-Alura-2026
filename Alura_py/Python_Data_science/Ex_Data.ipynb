```python
notas = []

for i in range(1,6):
    nota = int(input(f'Digite a {i}º nota do skatista'))
    notas.append(nota)

def media():
    notas.remove(max(notas))
    notas.remove(min(notas))
    return sum(notas)/len(notas)

print(f'\nMédia das notas do skatista: {media():.2f}')
```

```python
notas = []

for i in range(1,5):
    nota = int(input(f'Digite a {i}º nota do(a) estudante: '))
    notas.append(nota)
    
def analise(lista_notas):
    media = sum(lista_notas) / len(lista_notas)
    maior = max(lista_notas)
    menor = min(lista_notas)
    if media >= 6:
        situacao = 'Aprovado(a)' 
    else:
        situacao = 'Reprovado(a)'
    return media,maior,menor,situacao

media,maior,menor,situacao = analise(notas)
print(f'O(a) estudante obteve uma média de {media}, com a sua maior nota de {maior} pontos e a menor nota de {menor} pontos e foi {situacao}')
```

```python
def calcula_pontos(gols_marcados, gols_sofridos):
    pontuacao = 0
    for i in range(len(gols_marcados)):
        if gols_marcados[i] > gols_sofridos[i]:
            pontuacao += 3
        elif gols_marcados [i] == gols_sofridos [i]:
            pontuacao += 1
    total_jogos = len(gols_marcados)
    pontos_possiveis = total_jogos * 3
    aproveitamento = (pontuacao / pontos_possiveis) * 100
    return aproveitamento, pontuacao

aproveitamento, pontuacao = calcula_pontos([2, 1, 3, 1, 0], [1, 2, 2, 1, 3])
print(f'Pontuação do time: {pontuacao}\nAproveitamento do time: {aproveitamento:.2f}%')
```

```python
cidade = input('Digite para qual cidade gostaria de viajar (Salvador, Fortaleza, Natal ou Aracaju): ').upper()
while cidade not in ('SALVADOR','FORTALEZA','NATAL','ARACAJU'):
    cidade = input('Digite um destino válido (Salvador, Fortaleza, Natal ou Aracaju): ').upper()
dias = int(input('Digite quantas diárias gostaria: '))
distancias = [850, 800, 300, 550]
passeios = [200, 400, 250, 300]
gasolina = 5
km_l = 14

def gasto_hotel(dias):
    return 150 * dias

def gasto_gasolina(cidade):
    if cidade == 'SALVADOR':
        return (2 * distancias[0] / km_l) * gasolina
    elif cidade == 'FORTALEZA':
        return (2 * distancias[1] / km_l) * gasolina
    elif cidade == 'NATAL':
        return (2 * distancias[2] / km_l) * gasolina
    elif cidade == 'ARACAJU':
        return (2 * distancias[3] / km_l) * gasolina

def gasto_passeio(cidade, dias):
    if cidade == 'SALVADOR':
        return passeios[0] * dias
    elif cidade == 'FORTALEZA':
        return passeios[1] * dias
    elif cidade == 'NATAL':
        return passeios[2] * dias
    elif cidade == 'ARACAJU':
        return passeios[3] * dias

gastos = gasto_hotel(dias) + gasto_gasolina(cidade) + gasto_passeio(cidade, dias)



print(f'\nCom base nos gastos definidos, uma viagem de {dias} dias para {cidade.capitalize()} saindo de Recife custaria R$ {gastos:.2f}')
```
